## Input: grid-puzzles-ordered.csv

## Output: grid-puzzles.csv [ has uuid for merge operations ]

```python
import pandas as pd 
from uuid import uuid4

df = pd.read_csv('grid-puzzles-ordered.csv')
```

```python
for index, row in df.iterrows():
    # Generate a unique key for each row
    unique_key = str(uuid4())
    
    # Add the unique key to the DataFrame
    df.at[index, 'key'] = unique_key

df.to_csv('grid-puzzles.csv', index=False)
```

```python
df.shape[0] == df['key'].nunique()
```
