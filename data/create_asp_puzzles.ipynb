```python
import pandas as pd

raw_puzzles = pd.read_csv("grid-puzzles.csv")
raw_puzzles.head()
```

```python
partial = pd.read_csv("partial_solution.csv")
partial.head(n=6)
```

```python
raw_puzzles = raw_puzzles[["Puzzle"]]
raw_puzzles['Puzzle'] = raw_puzzles['Puzzle'].str.strip()

partial['Puzzle'] = partial['Puzzle'].str.strip()
merged = pd.merge(raw_puzzles, partial, on=["Puzzle"], how="left")
merged.head()
```

```python
import hashlib
from uuid import uuid4

for index, row in merged.iterrows():
    # Generate an MD5 hash
    unique_key = hashlib.md5(str(row['Puzzle']).encode()).hexdigest()
    merged.at[index, 'Key'] = unique_key

merged.to_csv('asp-tbd.csv', index=False)
```

```python
filled = 1 - merged['Code'].isna().sum()/len(merged)
print(f'Filled Pct: {round(filled*100,2)}%')
```
