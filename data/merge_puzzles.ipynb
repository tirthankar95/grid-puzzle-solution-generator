```python
def getKey(x: str):
    import hashlib 
    x = str(x).strip().lower()
    return hashlib.md5(x.encode('utf-8')).hexdigest()
```

```python
import pandas as pd 

main_puzzle = pd.read_csv("grid-puzzles-sample.csv")
main_puzzle['Key'] = main_puzzle['Puzzle'].apply(lambda x: getKey(x))
print(main_puzzle.shape)
main_puzzle.head()
```

```python
asp_solution = pd.read_csv("gflash25/grid-puzzles-sample-asp.csv")
asp_solution['Key'] = asp_solution["Puzzle"].apply(lambda x: getKey(x))
print(asp_solution.shape)
asp_solution.head()
```

```python
df = pd.merge(main_puzzle, asp_solution, on=["Key", "Puzzle"])
df = df[['Key', 'Puzzle', 'Code', 'Solution']]
prev_shape = df.shape
df_good = df[df['Solution'] != 'Incorrect verification code']
df_bad = df[df['Solution'] == 'Incorrect verification code']
df_good.to_csv('dataset.csv', index=False)
print(f'Before: {prev_shape},\nGood: {df_good.shape},\nBad: {df_bad.shape}')
```
