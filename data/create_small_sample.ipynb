```python
import pandas as pd

df = pd.read_csv("grid-puzzles-ordered.csv")
df.head()
```

```python
N_SAMPLES = 2
df_sample = df.groupby(['Grid', 'Difficulty']).sample(n=N_SAMPLES, random_state=42)\
    .reset_index(drop=True)

assert df_sample.groupby(['Grid', 'Difficulty']).size().max() == N_SAMPLES and \
    df_sample.groupby(['Grid', 'Difficulty']).size().min() == N_SAMPLES, \
    "N_SAMPLES per group"
```

```python
df_sample.to_csv("grid-puzzles-sample.csv", index=False)
```
